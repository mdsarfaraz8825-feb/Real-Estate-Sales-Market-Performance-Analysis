```python
import pandas as pd 
import numpy as np
```

```python
agents = pd.read_csv("cleaned_data/agents_cleaned.csv")
customers = pd.read_csv("cleaned_data/customers_cleaned.csv")
properties = pd.read_csv("cleaned_data/properties_cleaned.csv")
sales = pd.read_csv("cleaned_data/sales_cleaned.csv")
locations = pd.read_csv("cleaned_data/locations_cleaned.csv")
amenities = pd.read_csv("cleaned_data/amenities_cleaned.csv")
```

```python
for name, df in {
    "agents": agents,
    "customers": customers,
    "properties": properties,
    "sales": sales,
    "locations": locations,
    "amenities": amenities
}.items():
    print(f"\n{name}")
    print(df.shape)
    print(df.info())
```

```python
agents = agents.drop_duplicates(subset="Agent_ID")
customers = customers.drop_duplicates(subset="Customer_ID")
properties = properties.drop_duplicates(subset="Property_ID")
sales = sales.drop_duplicates(subset="Sale_ID")
```

```python
agents["Experience_Years"] = pd.to_numeric(
    agents["Experience_Years"],
    errors="coerce"
)

agents["Commission_Rate"] = pd.to_numeric(
    agents["Commission_Rate"],
    errors="coerce"
)

properties["Construction_Year"] = pd.to_numeric(
    properties["Construction_Year"],
    errors="coerce"
)

properties["Listing_Price"] = pd.to_numeric(
    properties["Listing_Price"],
    errors="coerce"
)

properties["Selling_Price"] = pd.to_numeric(
    properties["Selling_Price"],
    errors="coerce"
)
```

```python
sales["Listing_Date"] = pd.to_datetime(
    sales["Listing_Date"]
)

sales["Sale_Date"] = pd.to_datetime(
    sales["Sale_Date"]
)
```

```python
agents["Gender"] = (
    agents["Gender"]
    .str.strip()
    .str.title()
)

customers["Gender"] = (
    customers["Gender"]
    .str.strip()
    .str.title()
)

customers["Marital_Status"] = (
    customers["Marital_Status"]
    .str.strip()
    .str.title()
)

properties["Property_Type"] = (
    properties["Property_Type"]
    .str.strip()
    .str.title()
)
```

```python
agents["Commission_Rate"] = agents["Commission_Rate"].fillna(
    agents["Commission_Rate"].mean()
)

properties["Bathrooms"] = properties["Bathrooms"].fillna(
    properties["Bathrooms"].median()
)
```

```python
datasets = {
    "Agents": agents,
    "Customers": customers,
    "Properties": properties,
    "Sales": sales,
    "Locations": locations,
    "Amenities": amenities
}

for name, df in datasets.items():
    print(name, df.shape)
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
assert agents["Agent_ID"].isnull().sum() == 0
assert customers["Customer_ID"].isnull().sum() == 0
assert properties["Property_ID"].isnull().sum() == 0
assert sales["Sale_ID"].isnull().sum() == 0
```

```python
assert agents["Agent_ID"].duplicated().sum() == 0
assert properties["Property_ID"].duplicated().sum() == 0
```

```python
agents.to_csv("../data/agents_cleaned.csv", index=False)

customers.to_csv("../data/customers_cleaned.csv", index=False)

properties.to_csv("../data/properties_cleaned.csv", index=False)

sales.to_csv("../data/sales_cleaned.csv", index=False)

locations.to_csv("../data/locations_cleaned.csv", index=False)

amenities.to_csv("../data/amenities_cleaned.csv", index=False)
```
